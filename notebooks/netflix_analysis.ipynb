# Netflix Content Explorer

This project explores the Netflix Titles dataset using Python, Pandas, and NumPy.

### Objectives
- Verify the dataset and the environment
- Clean and prepare the data
- Apply the required analysis operations
- Build a working menu-driven application


## 1. Environment and dataset verification

The notebook uses the project virtual environment and reads the dataset from the repository `dataset/` folder.


```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np

candidates = [Path.cwd().resolve(), Path.cwd().resolve().parent]
project_root = None
for candidate in candidates:
    if (candidate / "dataset").exists():
        project_root = candidate
        break

if project_root is None:
    raise FileNotFoundError("Could not locate the project dataset folder")

if (project_root / "src").exists():
    sys.path.insert(0, str(project_root / "src"))

dataset_dir = project_root / "dataset"
dataset_files = sorted(dataset_dir.iterdir(), key=lambda p: p.name.lower())
dataset_files = [p for p in dataset_files if p.suffix.lower() in {'.csv', '.xlsx', '.xls'}]
dataset_path = dataset_files[0] if dataset_files else None
if dataset_path is None:
    raise FileNotFoundError("No dataset found in dataset/ folder")

df = pd.read_csv(dataset_path) if dataset_path.suffix.lower() == '.csv' else pd.read_excel(dataset_path)
print(f"Dataset file: {dataset_path.name}")
print(f"Rows: {len(df)} | Columns: {len(df.columns)}")
print("Columns:", list(df.columns))
print(df.head(3).to_string(index=False))
```

## 2. Data cleaning

The original dataset is preserved. The notebook works on a cleaned copy for the required analysis.


```python
from netflix_analysis import clean_dataset

cleaned_df, diagnostics = clean_dataset(df)
print('Rows before cleaning:', diagnostics['rows_before_cleaning'])
print('Rows after cleaning:', diagnostics['rows_after_cleaning'])
print('Invalid date values:', len(diagnostics['invalid_date_values']))
print(cleaned_df.head(3).to_string(index=False))
```

## 3. Required analysis operations

This section demonstrates every required calculation using real dataset outputs.


```python
from netflix_analysis import (
    count_movies_vs_shows,
    titles_added_each_year,
    top_producing_countries,
    top_genres,
    rating_distribution,
    oldest_and_newest_release_year,
    search_titles,
    content_added_in_year,
    count_unique_countries,
    release_year_summary,
)

print('Movies vs TV Shows\n')
print(count_movies_vs_shows(cleaned_df))
print('\nTitles added each year (head)\n')
print(titles_added_each_year(cleaned_df).head(10))
print('\nTop 10 countries\n')
print(top_producing_countries(cleaned_df, 10))
print('\nTop 5 genres\n')
print(top_genres(cleaned_df, 5))
print('\nRating distribution\n')
print(rating_distribution(cleaned_df).head(10))
print('\nOldest and newest release years')
print(oldest_and_newest_release_year(cleaned_df))
print('\nKeyword search results for "love"')
print(search_titles(cleaned_df, 'love').head(5)[['title', 'type']])
print('\nTitles added in 2021')
print(content_added_in_year(cleaned_df, 2021).head(5)[['title', 'year_added']])
print('\nUnique countries count')
print(count_unique_countries(cleaned_df))
print('\nRelease year summary')
print(release_year_summary(cleaned_df))
```

## 4. Insights

1. Movies are much more common than TV shows in the final dataset.
2. Most titles were added in the late 2010s, showing strong growth in streaming content.
3. The United States is the dominant content producer by a large margin.
